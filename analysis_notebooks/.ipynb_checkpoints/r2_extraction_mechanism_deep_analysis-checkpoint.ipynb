# Deep Analysis of Block 2 Position Extraction Mechanism

This notebook provides a comprehensive investigation into **exactly** how Block 2's attention extracts position information in the R2 model variant.

## Key Questions
1. **What does the linear head learn?** - Which dimensions are most important?
2. **How do value vectors behave?** - Variance/norm dynamics across positions
3. **What's the mathematical mechanism?** - Exact decomposition of the computation
4. **Why does attention output have position signal?** - What creates the -0.93 correlation?

## Model: R2 Post-Attn Head (ff0hgn5g)
- Block 1: Frozen random weights (Xavier init)
- Block 2: Trained attention only (1 head)
- Head: Linear probe on post-attention residual stream

```python
# =========================
# Plotting utilities
# =========================

from pathlib import Path
from datetime import datetime
import plotly.express as px
import plotly.graph_objects as go
from transformer_lens.utils import to_numpy

try:
    BASE_DIR = Path(__file__).resolve().parent
except NameError:
    BASE_DIR = Path.cwd()

PLOTS_DIR = BASE_DIR / "overleaf" / "plots"
PLOTS_DIR.mkdir(parents=True, exist_ok=True)

def _safe_filename(title: str, fallback: str):
    if title:
        return title.lower().replace(" ", "_").replace("/", "_")
    return fallback

def line(
    tensor,
    line_labels=None,
    yaxis="",
    xaxis="",
    title="",
    legend_title="",
    show_legend=False,
    export=False,
    file_name=None,
    save_png=True,
    save_pdf=False,
    **kwargs,
):
    tensor = to_numpy(tensor)
    fig = px.line(tensor, template="plotly_white", **kwargs)
    fig.update_layout(
        title=dict(text=title, font=dict(size=24, family="Serif")),
        xaxis=dict(title=xaxis, title_font=dict(size=18), tickfont=dict(size=16)),
        yaxis=dict(title=yaxis, title_font=dict(size=18), tickfont=dict(size=16)),
        legend=dict(title=legend_title, font=dict(size=16)),
        showlegend=show_legend,
        width=800,
        height=500,
        margin=dict(l=50, r=50, t=80, b=50),
    )
    for trace in fig.data:
        trace.line.width = 3
        trace.marker = dict(symbol="circle", size=8)
        trace.showlegend = show_legend
    if line_labels:
        for i, label in enumerate(line_labels):
            fig.data[i].name = label
    if export:
        if file_name:
            base = PLOTS_DIR / file_name
        else:
            fname = _safe_filename(title, "line_plot")
            base = PLOTS_DIR / fname
        if save_png:
            fig.write_image(f"{base}.png", scale=3)
        if save_pdf:
            fig.write_image(f"{base}.pdf")
    fig.show()

def imshow(
    tensor,
    yaxis="",
    xaxis="",
    title="Image",
    export=False,
    file_name=None,
    save_png=True,
    save_pdf=False,
    col_markers=None,
    **kwargs,
):
    tensor = to_numpy(tensor)
    fig = px.imshow(
        tensor,
        color_continuous_scale="RdBu_r",
        color_continuous_midpoint=0.0,
        labels={"x": xaxis, "y": yaxis},
        template="plotly_white",
        aspect="equal",
        **kwargs,
    )
    fig.update_layout(
        title=dict(text=title, font=dict(size=24, family="Serif")),
        xaxis=dict(title=xaxis, title_font=dict(size=18), tickfont=dict(size=16)),
        yaxis=dict(title=yaxis, title_font=dict(size=18), tickfont=dict(size=16)),
        width=800,
        height=600,
        margin=dict(l=50, r=50, t=80, b=50),
    )
    if col_markers:
        y_marker = -0.6
        for item in col_markers:
            if isinstance(item, (int, float)):
                x, label, color = item, None, '#C44E52'
            else:
                x, label, color = item
            fig.add_trace(
                go.Scatter(
                    x=[x], y=[y_marker], mode='markers',
                    marker=dict(symbol='triangle-down', size=10, color=color),
                    showlegend=False, cliponaxis=False,
                )
            )
        fig.update_yaxes(range=[tensor.shape[0] - 0.5, -2.2])
    if export:
        if file_name:
            base = PLOTS_DIR / file_name
        else:
            fname = _safe_filename(title, "imshow")
            base = PLOTS_DIR / fname
        if save_png:
            fig.write_image(f"{base}.png", scale=3)
        if save_pdf:
            fig.write_image(f"{base}.pdf")
    fig.show()
```

```python
import sys
from pathlib import Path
from typing import Dict, List, Tuple, Optional

import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
from scipy import stats

torch.manual_seed(42)
np.random.seed(42)

device = "cuda" if torch.cuda.is_available() else "cpu"
BOS_TOKEN_ID = 50256

def find_repo_root(start: Path) -> Path:
    for parent in [start] + list(start.parents):
        if (parent / "nanoGPT").exists():
            return parent
    raise FileNotFoundError("Could not locate repo root containing nanoGPT/")

ROOT_DIR = find_repo_root(Path.cwd())
NANOGPT_DIR = ROOT_DIR / "nanoGPT"
sys.path.insert(0, str(NANOGPT_DIR))

from model_2layer_mechanism import TwoLayerMechanismModel, TwoLayerMechanismConfig

CHECKPOINT_PATH = ROOT_DIR / "nanoGPT/out-2layer-mechanism-r2-1head-postattn/R2/ff0hgn5g/best_ckpt.pt"
DATA_DIR = ROOT_DIR / "nanoGPT/data/openwebtext"

BATCH_SIZE = 64
VOCAB_SAMPLE_SIZE = 4096

print(f"Device: {device}")
print(f"Repo root: {ROOT_DIR}")
print(f"Checkpoint: {CHECKPOINT_PATH}")
```

```python
# =========================
# Helper Functions
# =========================

def load_r2_model(checkpoint_path: Path, device: str) -> TwoLayerMechanismModel:
    if not checkpoint_path.exists():
        raise FileNotFoundError(f"Checkpoint not found: {checkpoint_path}")
    checkpoint = torch.load(str(checkpoint_path), map_location=device, weights_only=False)
    config_dict = checkpoint["config"]
    config_fields = ["block_size", "vocab_size", "n_embd", "n_head", "dropout", "norm_type"]
    filtered = {k: config_dict[k] for k in config_fields if k in config_dict}
    filtered["bias"] = True
    filtered["use_regression"] = True
    config = TwoLayerMechanismConfig(**filtered)
    model = TwoLayerMechanismModel(config)
    model.set_post_attn_head(True)
    state_dict = {k.replace("_orig_mod.", ""): v for k, v in checkpoint["model"].items()}
    model.load_state_dict(state_dict)
    model.to(device)
    model.eval()
    return model

def load_owt_train_data(data_dir: Path) -> np.ndarray:
    train_path = data_dir / "train.bin"
    if not train_path.exists():
        raise FileNotFoundError(f"Train data not found: {train_path}")
    return np.memmap(str(train_path), dtype=np.uint16, mode="r")

def get_batch(data: np.ndarray, batch_size: int, block_size: int, device: str) -> torch.Tensor:
    tokens_needed = block_size - 1
    ix = torch.randint(len(data) - tokens_needed, (batch_size,))
    sequences = []
    for i in ix:
        i = i.item()
        after_bos = data[i : i + tokens_needed].astype(np.int64)
        seq = np.concatenate([[BOS_TOKEN_ID], after_bos])
        sequences.append(torch.from_numpy(seq))
    x = torch.stack(sequences)
    return x.to(device)

def make_positions(batch_size: int, block_size: int, device: str) -> torch.Tensor:
    return torch.arange(block_size, device=device).unsqueeze(0).repeat(batch_size, 1).float()

def corrcoef(x: torch.Tensor, y: torch.Tensor) -> float:
    x_flat = x.flatten().float()
    y_flat = y.flatten().float()
    if x_flat.numel() < 2:
        return 0.0
    x_centered = x_flat - x_flat.mean()
    y_centered = y_flat - y_flat.mean()
    denom = x_centered.norm() * y_centered.norm()
    if denom.item() == 0:
        return 0.0
    return (x_centered @ y_centered / denom).item()

def r2_score(preds: torch.Tensor, targets: torch.Tensor) -> float:
    preds = preds.flatten().float()
    targets = targets.flatten().float()
    ss_res = ((targets - preds) ** 2).sum()
    ss_tot = ((targets - targets.mean()) ** 2).sum()
    return (1 - ss_res / ss_tot).item() if ss_tot > 0 else 0.0

def linear_fit_r2(X: torch.Tensor, y: torch.Tensor) -> Tuple[float, np.ndarray]:
    X_np = X.detach().cpu().numpy()
    y_np = y.detach().cpu().numpy().reshape(-1)
    if X_np.ndim == 1:
        X_np = X_np.reshape(-1, 1)
    X_aug = np.concatenate([X_np, np.ones((X_np.shape[0], 1))], axis=1)
    w, *_ = np.linalg.lstsq(X_aug, y_np, rcond=None)
    y_hat = X_aug @ w
    ss_res = np.sum((y_np - y_hat) ** 2)
    ss_tot = np.sum((y_np - y_np.mean()) ** 2)
    r2 = 1 - ss_res / ss_tot if ss_tot > 0 else 0.0
    return float(r2), w

@torch.no_grad()
def compute_decoding_vector(model: TwoLayerMechanismModel, chunk_size: int = 4096) -> torch.Tensor:
    E = model.wte.weight.detach()
    w = torch.zeros(E.shape[1], device=E.device)
    for chunk in torch.split(E, chunk_size, dim=0):
        ln_chunk = model.block2.ln_1(chunk)
        w += ln_chunk.sum(dim=0)
    w = w / (w.norm() + 1e-8)
    return w

def cos_sim(v1: torch.Tensor, v2: torch.Tensor) -> float:
    v1 = v1.flatten().float()
    v2 = v2.flatten().float()
    return (F.normalize(v1.unsqueeze(0), dim=-1) @ F.normalize(v2.unsqueeze(0), dim=-1).T).item()

def summarize_position_signal(name: str, acts: torch.Tensor, positions: torch.Tensor) -> Dict[str, float]:
    norms = acts.norm(dim=-1)
    corr = corrcoef(norms, positions)
    r2, _ = linear_fit_r2(norms.flatten(), positions.flatten())
    print(f"{name}: corr(norm,pos)={corr:.4f}, R^2(norm->pos)={r2:.4f}")
    return {"corr": corr, "r2": r2}
```

```python
# =========================
# Load Model and Data
# =========================

model = load_r2_model(CHECKPOINT_PATH, device)
train_data = load_owt_train_data(DATA_DIR)
block_size = model.config.block_size
tokens = get_batch(train_data, BATCH_SIZE, block_size, device)
positions = make_positions(BATCH_SIZE, block_size, device)

with torch.no_grad():
    output, _ = model(tokens, capture_taps=True)
    taps = model.get_all_taps()
    attn1, attn2 = model.get_attention_weights()

D = model.config.n_embd
T = block_size
n_head = model.config.n_head
head_dim = D // n_head

# Extract weight matrices
W_Q2, W_K2, W_V2 = model.block2.attn.c_attn.weight.detach().split(D, dim=0)
b2 = model.block2.attn.c_attn.bias
if b2 is not None:
    b_Q2, b_K2, b_V2 = b2.detach().split(D, dim=0)
else:
    b_Q2 = b_K2 = b_V2 = None
W_O2 = model.block2.attn.c_proj.weight.detach()
b_O2 = model.block2.attn.c_proj.bias.detach() if model.block2.attn.c_proj.bias is not None else None

# Baseline R^2
preds = output.squeeze(-1)
baseline_r2 = r2_score(preds, positions)
print(f"\n=== BASELINE MODEL R^2: {baseline_r2:.4f} ===")
print("\n--- Position signal at each layer ---")
summarize_position_signal("embeddings", taps["embeddings"], positions)
summarize_position_signal("block1_out", taps["block1_out"], positions)
summarize_position_signal("block2_ln1", taps["block2_ln1"], positions)
summarize_position_signal("block2_attn", taps["block2_attn"], positions)
summarize_position_signal("block2_post_attn", taps["block2_post_attn"], positions)
```

## Section 1: Linear Head Weight Dissection

The pos_head is a single linear layer: `pred = w · LN(r2) + b`

Questions:
- Which dimensions have the largest weights?
- Does the head align with known subspaces (mean embedding, value SVD)?
- How sparse can we make the head while maintaining R²?

```python
# Extract head weights
w_head = model.pos_head.weight.detach().cpu().squeeze()  # [D]
b_head = model.pos_head.bias.detach().cpu().item() if model.pos_head.bias is not None else 0.0

print(f"Head weight shape: {w_head.shape}")
print(f"Head bias: {b_head:.4f}")
print(f"Weight norm: {w_head.norm():.4f}")
print(f"Weight mean: {w_head.mean():.4f}")
print(f"Weight std: {w_head.std():.4f}")

# Top 50 weights by magnitude
mag = w_head.abs()
top_vals, top_idx = mag.topk(50)

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.bar(range(50), top_vals.numpy())
plt.title("Top 50 pos_head weight magnitudes")
plt.xlabel("Rank")
plt.ylabel("|weight|")

plt.subplot(1, 2, 2)
plt.hist(w_head.numpy(), bins=100, edgecolor='black', alpha=0.7)
plt.title("Distribution of all head weights")
plt.xlabel("Weight value")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

print(f"\nTop 10 dimensions by magnitude: {top_idx[:10].tolist()}")
print(f"Top 10 weights: {w_head[top_idx[:10]].tolist()}")
```

```python
# Alignment with key subspaces
E = model.wte.weight.detach().cpu()
e_mean = E.mean(dim=0)
e_sum_ln = compute_decoding_vector(model).cpu()

print("=== Head Alignment with Key Directions ===")
print(f"cos(w_head, mean_embedding):     {cos_sim(w_head, e_mean):.4f}")
print(f"cos(w_head, sum_LN(embeddings)): {cos_sim(w_head, e_sum_ln):.4f}")

# Alignment with W_O @ W_V SVD components
B_val = (W_O2 @ W_V2).detach().cpu()  # [D, D]
U_val, S_val, Vh_val = torch.linalg.svd(B_val)

print("\n=== Alignment with W_O @ W_V SVD Components ===")
for i in range(10):
    sim = cos_sim(w_head, Vh_val[i])
    print(f"  Component {i} (σ={S_val[i]:.2f}): cos = {sim:.4f}")

# Alignment with residual stream covariance SVD
r2_out = taps["block2_post_attn"].reshape(-1, D).cpu()
r2_centered = r2_out - r2_out.mean(dim=0)
cov = (r2_centered.T @ r2_centered) / (r2_centered.shape[0] - 1)
U_cov, S_cov, Vh_cov = torch.linalg.svd(cov)

print("\n=== Alignment with Residual Stream Covariance SVD ===")
for i in range(10):
    sim = cos_sim(w_head, U_cov[:, i])
    print(f"  PC {i} (var={S_cov[i]:.2f}): cos = {sim:.4f}")
```

```python
# Sparse head test: Zero out all but top-k weights
print("=== Sparse Head Test: R² vs Number of Active Dimensions ===")

k_values = [5, 10, 20, 50, 100, 200, 300, 400, 500, 600, 700, 768]
sparse_r2_results = []

x_input = model.ln_f(taps["block2_post_attn"]).to(device)  # [B, T, D]
w_head_gpu = w_head.to(device)

for k in k_values:
    # Create sparse weight vector
    _, top_k_idx = mag.topk(k)
    w_sparse = torch.zeros_like(w_head_gpu)
    w_sparse[top_k_idx.to(device)] = w_head_gpu[top_k_idx.to(device)]
    
    # Compute predictions with sparse weights
    preds_sparse = (x_input @ w_sparse + b_head).squeeze(-1)
    r2_sparse = r2_score(preds_sparse, positions)
    sparse_r2_results.append(r2_sparse)
    print(f"  k={k:4d}: R² = {r2_sparse:.4f}")

plt.figure(figsize=(10, 5))
plt.plot(k_values, sparse_r2_results, 'o-', linewidth=2, markersize=8)
plt.axhline(y=baseline_r2, color='r', linestyle='--', label=f'Full head R²={baseline_r2:.4f}')
plt.axhline(y=0.95, color='g', linestyle=':', alpha=0.7, label='95% threshold')
plt.xlabel('Number of active dimensions (k)', fontsize=14)
plt.ylabel('R²', fontsize=14)
plt.title('Sparse Head: R² vs Active Dimensions', fontsize=16)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

# Find minimal k for R² > 0.95
for k, r2 in zip(k_values, sparse_r2_results):
    if r2 >= 0.95:
        print(f"\nMinimal k for R² > 0.95: {k}")
        break
```

## Section 2: Value Vector Dynamics Analysis

The attention mechanism computes:
- `v_t = W_V @ LN(r1_t)` - value vectors for each position
- `a_t = Σ_j α_tj v_j` - attention-weighted average

Key hypothesis: Position information emerges from averaging random vectors.
As position i increases, we average over more vectors, and the norm decreases.

```python
# Compute value vectors
ln1_out = taps["block2_ln1"]  # [B, T, D]
v_t = ln1_out @ W_V2.T  # [B, T, D]
if b_V2 is not None:
    v_t = v_t + b_V2

# Statistics per position
v_norm_per_pos = v_t.norm(dim=-1).mean(dim=0).cpu().numpy()  # [T]
v_mean_per_pos = v_t.mean(dim=0).cpu()  # [T, D]
v_std_per_pos = v_t.std(dim=0).mean(dim=-1).cpu().numpy()  # [T]

print(f"Value vector shape: {v_t.shape}")
print(f"Mean value norm: {v_norm_per_pos.mean():.4f}")
print(f"Value norm range: [{v_norm_per_pos.min():.4f}, {v_norm_per_pos.max():.4f}]")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].plot(v_norm_per_pos)
axes[0].set_title('Mean Value Vector Norm ||v_t|| vs Position')
axes[0].set_xlabel('Position t')
axes[0].set_ylabel('||v_t||')

axes[1].plot(v_std_per_pos)
axes[1].set_title('Std of Value Vectors vs Position')
axes[1].set_xlabel('Position t')
axes[1].set_ylabel('Std')

# Mean value vector norm per dimension
v_mean_norm = v_mean_per_pos.norm(dim=-1).numpy()
axes[2].plot(v_mean_norm)
axes[2].set_title('Norm of Mean Value Vector vs Position')
axes[2].set_xlabel('Position t')
axes[2].set_ylabel('||E[v_t]||')

plt.tight_layout()
plt.show()

# Correlation with position
pos_vec = torch.arange(T).float()
print(f"\nCorr(||v_t||, position): {corrcoef(torch.tensor(v_norm_per_pos), pos_vec):.4f}")
print(f"Corr(std(v_t), position): {corrcoef(torch.tensor(v_std_per_pos), pos_vec):.4f}")
```

```python
# Attention-weighted average: a_t = Σ_j α_tj v_j
# The attention output BEFORE W_O projection

attn_weights = attn2.squeeze(1)  # [B, T, T] (single head)
a_t = torch.einsum('bts,bsd->btd', attn_weights, v_t)  # [B, T, D]

a_norm_per_pos = a_t.norm(dim=-1).mean(dim=0).cpu().numpy()  # [T]
a_mean_per_pos = a_t.mean(dim=0).cpu()  # [T, D]

# After W_O projection (this is the attention output in the residual stream)
attn_out = taps["block2_attn"]  # [B, T, D]
attn_out_norm = attn_out.norm(dim=-1).mean(dim=0).cpu().numpy()  # [T]

print(f"Attention output shape (before W_O): {a_t.shape}")
print(f"Attention output shape (after W_O): {attn_out.shape}")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(a_norm_per_pos, label='Before W_O')
axes[0].plot(attn_out_norm, label='After W_O')
axes[0].set_title('Attention Output Norm vs Position')
axes[0].set_xlabel('Position t')
axes[0].set_ylabel('||a_t||')
axes[0].legend()

# Compare with theoretical 1/sqrt(t+1) decay
theoretical = a_norm_per_pos[0] / np.sqrt(np.arange(1, T+1))
axes[1].plot(a_norm_per_pos, label='Actual ||a_t||', linewidth=2)
axes[1].plot(theoretical, '--', label=r'$||a_0|| / \sqrt{t+1}$', alpha=0.7)
axes[1].set_title('Attention Output vs Theoretical Decay')
axes[1].set_xlabel('Position t')
axes[1].set_ylabel('Norm')
axes[1].legend()

plt.tight_layout()
plt.show()

print(f"\nCorr(||a_t|| before W_O, position): {corrcoef(torch.tensor(a_norm_per_pos), pos_vec):.4f}")
print(f"Corr(||a_t|| after W_O, position): {corrcoef(torch.tensor(attn_out_norm), pos_vec):.4f}")
```

## Section 3: The Averaging Mechanism (Mathematical Analysis)

**Core insight**: Position i averages over (i+1) random-ish vectors.

Under uniform causal attention:
- `a_i = (1/(i+1)) Σ_{j=0}^{i} v_j`
- If v_j are independent with E[v]=0, Var[v]=σ², then:
- `E[||a_i||²] ≈ σ²/(i+1)`

This creates a monotonic relationship between position and norm!

```python
# Compare actual attention with uniform causal attention

# Actual attention output norm
actual_a_norm = a_t.norm(dim=-1)  # [B, T]

# Uniform causal attention
uniform_attn = torch.zeros(T, T, device=device)
for i in range(T):
    uniform_attn[i, :i+1] = 1.0 / (i + 1)

# Compute attention output with uniform attention
a_t_uniform = torch.einsum('ts,bsd->btd', uniform_attn, v_t)  # [B, T, D]
uniform_a_norm = a_t_uniform.norm(dim=-1)  # [B, T]

# Compare
actual_mean = actual_a_norm.mean(dim=0).cpu().numpy()
uniform_mean = uniform_a_norm.mean(dim=0).cpu().numpy()

plt.figure(figsize=(10, 5))
plt.plot(actual_mean, label='Actual (learned) attention', linewidth=2)
plt.plot(uniform_mean, '--', label='Uniform causal attention', linewidth=2)
plt.xlabel('Position', fontsize=14)
plt.ylabel('||a_t|| (before W_O)', fontsize=14)
plt.title('Attention Output Norm: Learned vs Uniform', fontsize=16)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

# How close is learned attention to uniform?
attn_diff = (attn_weights - uniform_attn.unsqueeze(0)).abs().mean(dim=0).cpu().numpy()
print(f"Mean |learned - uniform| attention: {attn_diff.mean():.4f}")
print(f"Max |learned - uniform| attention: {attn_diff.max():.4f}")
```

```python
# Mathematical verification: E[||a_i||²] ≈ E[||v||²]/(i+1)

# Compute E[||v||²] - mean squared norm of value vectors
v_squared_norm_mean = (v_t.norm(dim=-1) ** 2).mean().item()
print(f"E[||v||²] = {v_squared_norm_mean:.4f}")

# Theoretical prediction for ||a_i||²
positions_arr = np.arange(1, T + 1)
theoretical_squared_norm = v_squared_norm_mean / positions_arr
theoretical_norm = np.sqrt(theoretical_squared_norm)

# Actual squared norms (with uniform attention for cleaner comparison)
actual_squared_norm = (uniform_a_norm ** 2).mean(dim=0).cpu().numpy()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(actual_squared_norm, label='Actual ||a_i||² (uniform attn)')
axes[0].plot(theoretical_squared_norm, '--', label=r'E[||v||²]/(i+1)', alpha=0.8)
axes[0].set_xlabel('Position i')
axes[0].set_ylabel('||a_i||²')
axes[0].set_title('Squared Norm: Actual vs Theory')
axes[0].legend()

# Log-log plot to see the 1/t relationship
axes[1].loglog(positions_arr, actual_squared_norm[0] * positions_arr / positions_arr[0], 
               label='Actual (normalized)')
axes[1].loglog(positions_arr, 1/positions_arr * positions_arr[0], '--', 
               label='1/t reference', alpha=0.8)
axes[1].set_xlabel('Position i (log)')
axes[1].set_ylabel('||a_i||² (log, normalized)')
axes[1].set_title('Log-Log Plot: Verifying 1/t Decay')
axes[1].legend()

plt.tight_layout()
plt.show()

# Fit slope in log-log space
log_pos = np.log(positions_arr[1:])  # skip pos 0
log_norm = np.log(actual_squared_norm[1:])
slope, intercept = np.polyfit(log_pos, log_norm, 1)
print(f"\nLog-log slope (should be ≈ -1 for 1/t): {slope:.4f}")
```

## Section 4: Per-Dimension Position Signal

## Section 5: Residual Stream Decomposition

The post-attention residual: `r2 = r1 + attn_out`

Questions:
- Which component carries position information?
- How do they combine?

```python
r1 = taps["block1_out"]  # [B, T, D]
attn_out = taps["block2_attn"]  # [B, T, D]
r2 = taps["block2_post_attn"]  # [B, T, D]

# Verify: r2 = r1 + attn_out
reconstruction_error = (r2 - (r1 + attn_out)).abs().max().item()
print(f"Reconstruction error (should be ~0): {reconstruction_error:.6f}")

# Position signal in each component
r1_norm = r1.norm(dim=-1)  # [B, T]
attn_norm = attn_out.norm(dim=-1)  # [B, T]
r2_norm = r2.norm(dim=-1)  # [B, T]

print(f"\n=== Position Signal by Component ===")
print(f"Corr(||r1||, pos):       {corrcoef(r1_norm, positions):.4f}")
print(f"Corr(||attn_out||, pos): {corrcoef(attn_norm, positions):.4f}")
print(f"Corr(||r2||, pos):       {corrcoef(r2_norm, positions):.4f}")

# R² from linear regression
r1_r2, _ = linear_fit_r2(r1_norm.flatten(), positions.flatten())
attn_r2, _ = linear_fit_r2(attn_norm.flatten(), positions.flatten())
r2_r2, _ = linear_fit_r2(r2_norm.flatten(), positions.flatten())

print(f"\nR²(||r1|| -> pos):       {r1_r2:.4f}")
print(f"R²(||attn_out|| -> pos): {attn_r2:.4f}")
print(f"R²(||r2|| -> pos):       {r2_r2:.4f}")
```

```python
# Orthogonal decomposition: Project r2 onto attn_out direction

# For each position, decompose r2 into component parallel to attn_out and orthogonal
def project_onto(v, u):
    """Project v onto direction of u"""
    u_norm = F.normalize(u, dim=-1)
    proj_scalar = (v * u_norm).sum(dim=-1, keepdim=True)
    proj_parallel = proj_scalar * u_norm
    proj_orthog = v - proj_parallel
    return proj_parallel, proj_orthog, proj_scalar.squeeze(-1)

r2_parallel, r2_orthog, proj_coef = project_onto(r2, attn_out)

print(f"Shape check:")
print(f"  r2_parallel: {r2_parallel.shape}")
print(f"  r2_orthog: {r2_orthog.shape}")
print(f"  proj_coef: {proj_coef.shape}")

# Position signal in decomposed components
parallel_norm = r2_parallel.norm(dim=-1)
orthog_norm = r2_orthog.norm(dim=-1)

print(f"\n=== Position Signal in Orthogonal Decomposition ===")
print(f"Corr(||r2_parallel||, pos): {corrcoef(parallel_norm, positions):.4f}")
print(f"Corr(||r2_orthog||, pos):   {corrcoef(orthog_norm, positions):.4f}")
print(f"Corr(proj_coef, pos):       {corrcoef(proj_coef, positions):.4f}")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].plot(parallel_norm.mean(dim=0).cpu().numpy(), label='Parallel to attn_out')
axes[0].plot(orthog_norm.mean(dim=0).cpu().numpy(), label='Orthogonal')
axes[0].set_xlabel('Position')
axes[0].set_ylabel('Norm')
axes[0].set_title('Decomposed r2 Norms')
axes[0].legend()

axes[1].plot(proj_coef.mean(dim=0).cpu().numpy())
axes[1].set_xlabel('Position')
axes[1].set_ylabel('Projection coefficient')
axes[1].set_title('r2 projection onto attn_out direction')

# Fraction of r2 norm from parallel component
frac_parallel = parallel_norm / (r2_norm + 1e-8)
axes[2].plot(frac_parallel.mean(dim=0).cpu().numpy())
axes[2].set_xlabel('Position')
axes[2].set_ylabel('Fraction')
axes[2].set_title('Fraction of ||r2|| from parallel component')

plt.tight_layout()
plt.show()
```

## Section 6: What the Head Actually Computes

The head computes: `pred = w · LN(r2) + b = w · LN(r1 + attn_out) + b`

Questions:
- What is the contribution from r1 vs attn_out?
- Can we replace r2 with just attn_out?

```python
# Apply final LN to different inputs
ln_r2 = model.ln_f(r2)  # Full input
ln_r1 = model.ln_f(r1)  # Just r1
ln_attn = model.ln_f(attn_out)  # Just attn_out

# Predictions from each
w_gpu = model.pos_head.weight.detach().squeeze()  # [D]
b_gpu = model.pos_head.bias.detach().item() if model.pos_head.bias is not None else 0.0

pred_full = (ln_r2 @ w_gpu + b_gpu).detach()  # [B, T]
pred_r1_only = (ln_r1 @ w_gpu + b_gpu).detach()  # [B, T]
pred_attn_only = (ln_attn @ w_gpu + b_gpu).detach()  # [B, T]

print("=== R² from Different Input Components ===")
print(f"Full r2 = r1 + attn_out: R² = {r2_score(pred_full, positions):.4f}")
print(f"r1 only:                 R² = {r2_score(pred_r1_only, positions):.4f}")
print(f"attn_out only:           R² = {r2_score(pred_attn_only, positions):.4f}")

# Visualize predictions
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sample_idx = 0
pos_true = positions[sample_idx].cpu().numpy()

axes[0].scatter(pos_true, pred_full[sample_idx].cpu().numpy(), alpha=0.5, s=10)
axes[0].plot([0, T], [0, T], 'r--', label='y=x')
axes[0].set_xlabel('True position')
axes[0].set_ylabel('Predicted')
axes[0].set_title(f'Full r2 (R²={r2_score(pred_full, positions):.4f})')

axes[1].scatter(pos_true, pred_r1_only[sample_idx].cpu().numpy(), alpha=0.5, s=10)
axes[1].plot([0, T], [0, T], 'r--', label='y=x')
axes[1].set_xlabel('True position')
axes[1].set_ylabel('Predicted')
axes[1].set_title(f'r1 only (R²={r2_score(pred_r1_only, positions):.4f})')

axes[2].scatter(pos_true, pred_attn_only[sample_idx].cpu().numpy(), alpha=0.5, s=10)
axes[2].plot([0, T], [0, T], 'r--', label='y=x')
axes[2].set_xlabel('True position')
axes[2].set_ylabel('Predicted')
axes[2].set_title(f'attn_out only (R²={r2_score(pred_attn_only, positions):.4f})')

plt.tight_layout()
plt.show()
```

```python
# Linear decomposition: How much does each component contribute?
# Note: LN is not linear, so this is approximate

# Compute "effective contribution" via gradients or approximation
# Simpler approach: regress predictions on component norms

from sklearn.linear_model import LinearRegression

X = np.column_stack([
    r1_norm.flatten().cpu().numpy(),
    attn_norm.flatten().cpu().numpy(),
    np.ones(BATCH_SIZE * T)
])
y = pred_full.flatten().cpu().numpy()

reg = LinearRegression(fit_intercept=False)
reg.fit(X, y)

print("=== Linear Decomposition of Predictions ===")
print(f"pred ≈ {reg.coef_[0]:.4f} * ||r1|| + {reg.coef_[1]:.4f} * ||attn_out|| + {reg.coef_[2]:.4f}")
print(f"Regression R²: {reg.score(X, y):.4f}")

# More detailed: use the actual activations, not just norms
ln_r2_flat = ln_r2.reshape(-1, D).detach().cpu().numpy()
w_np = w_gpu.cpu().numpy()

# Contribution of each dimension
contributions = ln_r2_flat * w_np
total_contribution = contributions.sum(axis=1)

# Variance of contribution by dimension
dim_contribution_var = contributions.var(axis=0)
top_contributing_dims = np.argsort(dim_contribution_var)[-20:][::-1]

print(f"\nTop 10 dimensions by contribution variance:")
for d in top_contributing_dims[:10]:
    print(f"  Dim {d}: var = {dim_contribution_var[d]:.4f}, weight = {w_np[d]:.4f}")
```

## Section 7: Synthetic Interventions

Test the mechanism with controlled inputs:
1. All-same-token sequences
2. Orthogonal embeddings
3. Scaled embeddings
4. Uniform attention swap

```python
# (a) Constant Token Input (except BOS)
print("=== Intervention 1: Constant Token Sequences ===")

const_tokens = [50, 100, 500, 1000, 5000, 10000, 20000, 40000]
const_r2_results = []

for const_token in const_tokens:
    x_const = torch.full_like(tokens, const_token)
    x_const[:, 0] = BOS_TOKEN_ID
    
    with torch.no_grad():
        preds_const, _ = model(x_const)
    
    r2_const = r2_score(preds_const.squeeze(), positions)
    const_r2_results.append(r2_const)
    print(f"  Token {const_token}: R² = {r2_const:.4f}")

print(f"\nMean R² with constant tokens: {np.mean(const_r2_results):.4f}")
print(f"Std R² with constant tokens:  {np.std(const_r2_results):.4f}")
```

```python
# (b) Shuffled tokens (BOS fixed at position 0)
print("=== Intervention 2: Shuffled Token Sequences ===")

def shuffle_tokens_except_bos(t):
    B, T = t.shape
    t_shuf = t.clone()
    for i in range(B):
        idx = torch.randperm(T-1, device=t.device) + 1
        t_shuf[i, 1:] = t[i, idx]
    return t_shuf

shuffle_r2_results = []
for _ in range(5):
    x_shuf = shuffle_tokens_except_bos(tokens)
    with torch.no_grad():
        preds_shuf, _ = model(x_shuf)
    r2_shuf = r2_score(preds_shuf.squeeze(), positions)
    shuffle_r2_results.append(r2_shuf)

print(f"R² with shuffled tokens (5 runs): {shuffle_r2_results}")
print(f"Mean: {np.mean(shuffle_r2_results):.4f}, Std: {np.std(shuffle_r2_results):.4f}")
```

```python
# (c) Uniform Causal Attention Swap
print("=== Intervention 3: Uniform Causal Attention Swap ===")

def forward_with_uniform_attn(model, r1, W_V, W_O, b_V, b_O):
    """Forward pass through Block 2 with uniform causal attention."""
    B, T, D = r1.shape
    
    # LN1
    ln1 = model.block2.ln_1(r1)
    
    # Values
    v = ln1 @ W_V.T
    if b_V is not None:
        v = v + b_V
    
    # Uniform causal attention
    uniform_attn = torch.zeros(T, T, device=r1.device)
    for i in range(T):
        uniform_attn[i, :i+1] = 1.0 / (i + 1)
    
    # Attention output
    a = torch.einsum('ts,bsd->btd', uniform_attn, v)
    attn_out = a @ W_O.T
    if b_O is not None:
        attn_out = attn_out + b_O
    
    # Post-attention residual
    post_attn = r1 + attn_out
    
    # Final LN + head
    x_f = model.ln_f(post_attn)
    preds = model.pos_head(x_f)
    
    return preds

r1 = taps["block1_out"]
preds_uniform = forward_with_uniform_attn(model, r1, W_V2, W_O2, b_V2, b_O2)
r2_uniform = r2_score(preds_uniform.squeeze(), positions)

print(f"R² with learned attention:  {baseline_r2:.4f}")
print(f"R² with uniform attention:  {r2_uniform:.4f}")
print(f"Difference:                 {baseline_r2 - r2_uniform:.4f}")
```

```python
# (d) Scaled Embeddings Test
# If position comes from averaging, scaling should not affect R² (LN normalizes)
print("=== Intervention 4: Scaled Embeddings ===")

# We can't easily modify embeddings in the model, but we can scale the input activations
scales = [0.1, 0.5, 1.0, 2.0, 5.0, 10.0]
scaled_r2_results = []

for scale in scales:
    # Scale the r1 (Block 1 output) and run Block 2
    r1_scaled = r1 * scale
    preds_scaled = forward_with_uniform_attn(model, r1_scaled, W_V2, W_O2, b_V2, b_O2)
    r2_scaled = r2_score(preds_scaled.squeeze(), positions)
    scaled_r2_results.append(r2_scaled)
    print(f"  Scale {scale:5.1f}: R² = {r2_scaled:.4f}")

print(f"\nR² is {'stable' if np.std(scaled_r2_results) < 0.01 else 'NOT stable'} across scales")
```

## Section 8: Causal Dimension Ablation

Find the minimal sufficient subspace for position decoding.
Zero out dimensions one by one and track R² degradation.

```python
# Identify top-k dimensions by head weight magnitude
x_input = model.ln_f(taps["block2_post_attn"]).to(device)
w_head_gpu = w_head.to(device)

# Order dimensions by weight magnitude (most important first)
sorted_dims = torch.argsort(w_head.abs(), descending=True)

# Progressive ablation: keep only top-k dimensions
print("=== Progressive Ablation: Keeping Top-k Dimensions ===")
k_values = [10, 25, 50, 75, 100, 150, 200, 300, 400, 500, 600, 700, 768]
keep_r2_results = []

for k in k_values:
    # Create mask for top-k dimensions
    mask = torch.zeros(D, device=device)
    mask[sorted_dims[:k].to(device)] = 1.0
    
    # Apply mask to weights
    w_masked = w_head_gpu * mask
    preds_masked = (x_input @ w_masked + b_head)
    r2_masked = r2_score(preds_masked, positions)
    keep_r2_results.append(r2_masked)
    print(f"  Keep top {k:4d}: R² = {r2_masked:.4f}")

# Reverse: ablate top-k dimensions (remove most important)
print("\n=== Reverse Ablation: Removing Top-k Dimensions ===")
remove_r2_results = []

for k in k_values[:-1]:  # Skip k=768 (would remove all)
    mask = torch.ones(D, device=device)
    mask[sorted_dims[:k].to(device)] = 0.0
    
    w_masked = w_head_gpu * mask
    preds_masked = (x_input @ w_masked + b_head)
    r2_masked = r2_score(preds_masked, positions)
    remove_r2_results.append(r2_masked)
    print(f"  Remove top {k:4d}: R² = {r2_masked:.4f}")

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(k_values, keep_r2_results, 'o-', linewidth=2, markersize=8)
axes[0].axhline(y=0.95, color='g', linestyle=':', alpha=0.7, label='95% threshold')
axes[0].set_xlabel('Number of dimensions kept')
axes[0].set_ylabel('R²')
axes[0].set_title('R² vs Dimensions Kept (by weight magnitude)')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(k_values[:-1], remove_r2_results, 'o-', linewidth=2, markersize=8, color='red')
axes[1].set_xlabel('Number of dimensions removed')
axes[1].set_ylabel('R²')
axes[1].set_title('R² vs Dimensions Removed (most important first)')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Individual dimension ablation: effect of removing each dimension
print("=== Individual Dimension Ablation (Top 20 by weight) ===")

individual_effects = []
for dim in sorted_dims[:20].tolist():
    mask = torch.ones(D, device=device)
    mask[dim] = 0.0
    
    w_masked = w_head_gpu * mask
    preds_masked = (x_input @ w_masked + b_head)
    r2_masked = r2_score(preds_masked, positions)
    
    effect = baseline_r2 - r2_masked
    individual_effects.append((dim, effect, w_head[dim].item()))
    print(f"  Remove dim {dim:4d}: R² = {r2_masked:.4f} (Δ = {effect:.4f}, weight = {w_head[dim].item():.4f})")

# Most impactful individual dimensions
individual_effects.sort(key=lambda x: -x[1])
print("\nMost impactful dimensions (by R² drop when removed):")
for dim, effect, weight in individual_effects[:5]:
    print(f"  Dim {dim}: ΔR² = {effect:.4f}, weight = {weight:.4f}")
```

## Section 9: Mathematical Formula Derivation

Derive a closed-form approximation for the position prediction.

Starting from: `pred_i = w · LN(r1_i + Σ_j α_ij W_O W_V LN(r1_j)) + b`

Under uniform attention: `α_ij = 1/(i+1)` for j ≤ i

```python
# Simplified formula under uniform attention
# pred_i ≈ w · LN(r1_i + (1/(i+1)) Σ_j≤i W_O W_V LN(r1_j)) + b

# Let's compute each component step by step
r1 = taps["block1_out"]  # [B, T, D]
B, T, D = r1.shape

# 1. LN(r1)
ln_r1 = model.block2.ln_1(r1)  # [B, T, D]

# 2. Value projection: W_V LN(r1)
v = ln_r1 @ W_V2.T  # [B, T, D]
if b_V2 is not None:
    v = v + b_V2

# 3. Cumulative sum for uniform averaging: Σ_j≤i v_j
v_cumsum = torch.cumsum(v, dim=1)  # [B, T, D]

# 4. Uniform average: (1/(i+1)) Σ_j≤i v_j
divisor = torch.arange(1, T+1, device=device).view(1, T, 1).float()
v_avg = v_cumsum / divisor  # [B, T, D]

# 5. Output projection: W_O @ v_avg
attn_out_formula = v_avg @ W_O2.T  # [B, T, D]
if b_O2 is not None:
    attn_out_formula = attn_out_formula + b_O2

# 6. Residual: r1 + attn_out
r2_formula = r1 + attn_out_formula  # [B, T, D]

# 7. Final LN + head
ln_r2_formula = model.ln_f(r2_formula)
pred_formula = (ln_r2_formula @ w_head_gpu + b_head)

# Compare with actual
r2_formula_score = r2_score(pred_formula, positions)
print(f"=== Formula-Based Prediction ===")
print(f"R² from derived formula (uniform attn): {r2_formula_score:.4f}")
print(f"R² from actual model:                   {baseline_r2:.4f}")
```

```python
# Even simpler: Can we predict position from just ||attn_out||
# Since ||attn_out|| ∝ 1/sqrt(i+1), and pred ∝ i, we have pred ∝ 1/||attn_out||²

attn_out_actual = taps["block2_attn"]
attn_out_norm = attn_out_actual.norm(dim=-1)  # [B, T]

# Fit: position = a / ||attn_out||² + c
inv_norm_sq = 1 / (attn_out_norm ** 2 + 1e-8)
r2_inv_norm, coeffs = linear_fit_r2(inv_norm_sq.flatten(), positions.flatten())

print(f"=== Simple Formula: pos ≈ a / ||attn_out||² + c ===")
print(f"Fitted coefficients: a = {coeffs[0]:.4f}, c = {coeffs[1]:.4f}")
print(f"R² from inverse squared norm: {r2_inv_norm:.4f}")

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Scatter: position vs 1/||attn_out||²
axes[0].scatter(inv_norm_sq.flatten().cpu().numpy(), 
                positions.flatten().cpu().numpy(), alpha=0.1, s=1)
# Fit line
x_fit = np.linspace(inv_norm_sq.min().item(), inv_norm_sq.max().item(), 100)
y_fit = coeffs[0] * x_fit + coeffs[1]
axes[0].plot(x_fit, y_fit, 'r-', linewidth=2, label=f'Linear fit (R²={r2_inv_norm:.3f})')
axes[0].set_xlabel('1 / ||attn_out||²')
axes[0].set_ylabel('Position')
axes[0].set_title('Position vs Inverse Squared Attention Norm')
axes[0].legend()

# Predicted vs actual
pred_simple = coeffs[0] * inv_norm_sq + coeffs[1]
axes[1].scatter(positions.flatten().cpu().numpy(),
                pred_simple.flatten().cpu().numpy(), alpha=0.1, s=1)
axes[1].plot([0, T], [0, T], 'r--', linewidth=2)
axes[1].set_xlabel('True position')
axes[1].set_ylabel('Predicted (simple formula)')
axes[1].set_title(f'Simple Formula Predictions (R²={r2_inv_norm:.3f})')

plt.tight_layout()
plt.show()
```
